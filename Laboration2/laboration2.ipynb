## Förbered datan: konvertera strängar till tal

När CSV-filen läses in blir även siffrorna strängar, till exempel `"26"`. För resten av laborationen behöver årsvärdena vara heltal.

**Kör cellen nedan. Du behöver inte kunna skriva eller förstå all kod i den ännu.** Den förbereder bara datan och sparar resultatet i `namndata_tal`.

````python
namndata_tal = []

for rad in namndata:
    ny_rad = [rad[0]]
    for ar in rad[1:]:
        ny_rad.append(int(ar))
    namndata_tal.append(ny_rad)
````
