# Ett litet problem

En bil kör **150 km** och använder **9 liter** bränsle.

Hur räknar vi ut liter per 100 km?


````python
distance = 150
fuel = 9

consumption = fuel / distance * 100
consumption
````

# Gör resultatet begripligt

Ett tal som `6.0` säger inte så mycket utan sammanhang.


````python
print(f"Förbrukning: {consumption} liter/100 km")
````

# En första glimt av villkor

Hittills har Python gjort samma sak varje gång.

Men ett program kan också **välja vad som ska hända**.


````python
temperature = 18

if temperature > 20:
    print("Det är varmt")
else:
    print("Det är inte särskilt varmt")
````
