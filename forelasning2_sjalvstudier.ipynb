## 2. Jämförelser

En jämförelse ger alltid ett booleskt värde: `True` eller `False`.

Vanliga operatorer:

| Operator | Betydelse |
|---|---|
| `<` | mindre än |
| `>` | större än |
| `<=` | mindre än eller lika med |
| `>=` | större än eller lika med |
| `==` | lika med |
| `!=` | inte lika med |

Observera skillnaden mellan `=` och `==`: `=` **tilldelar** ett värde, medan `==` **jämför** två värden.

````python
print(18 < 20)
print(18 > 20)
print(20 == 20)
print(20 != 20)
````

## 3. `if`: gör något om ett villkor är sant

En `if`-sats låter programmet välja om ett kodblock ska köras.

Indraget är en del av Python-syntaxen: raderna som hör till `if` måste vara indragna.

````python
temperature = 23

if temperature > 20:
    print("Det är varmt")
    print("Ta med vatten")

print("Programmet fortsätter här")
````

Ändra `temperature` till `18` och kör igen. Vilka rader skrivs ut nu?

## 4. `if` och `else`

Med `else` kan vi välja mellan två alternativ.

````python
temperature = 18

if temperature > 20:
    print("Det är varmt")
else:
    print("Det är inte varmt")
````

### `elif`: fler än två alternativ

Python testar villkoren **uppifrån och ned** och kör det första block vars villkor är sant.

````python
temperature = 15

if temperature < 10:
    print("Kallt")
elif temperature < 20:
    print("Lagom")
else:
    print("Varmt")
````

## 6. Varför loopar?

Om vi vill göra samma sak med varje värde i en lista vill vi inte skriva samma kodrad om och om igen.

````python
temperatures = [12, 15, 18, 17, 11]

for temperature in temperatures:
    print(temperature)
````

Läs loopen nästan som svenska:

> För varje `temperature` i `temperatures`: skriv ut `temperature`.

Loopvariabeln får ett nytt värde från listan varje varv.

### Prova själv

Vad skrivs ut? Förutsäg innan du kör.

````python
names = ["Ada", "Grace", "Guido"]

for name in names:
    print(f"Hej {name}!")
````

## 7. Loopen kan göra beräkningar

Vi kan använda det aktuella värdet i en beräkning.

````python
temperatures_c = [0, 10, 20, 30]

for celsius in temperatures_c:
    fahrenheit = celsius * 9 / 5 + 32
    print(f"{celsius} °C = {fahrenheit} °F")
````

## 8. Den viktiga kombinationen: `for` + `if`

Nu kan vi gå igenom flera värden och bara reagera på de värden som uppfyller ett villkor.

````python
prices = [18.95, 19.50, 20.10, 19.85, 21.30]

for price in prices:
    if price > 20:
        print(price)
````

Det här mönstret är centralt i laboration 2: **gå igenom → kontrollera villkor → gör något**.

## 9. Mönster 1: räkna

Om vi vill veta **hur många** värden som uppfyller ett villkor behöver vi en räknare.

Räknaren börjar på 0 och ökas när vi hittar ett värde vi vill räkna.

````python
count = 0

for price in prices:
    if price > 20:
        count = count + 1

print(f"Antal priser över 20 kr: {count}")
````

## 10. Mönster 2: summera

En summa byggs upp på nästan samma sätt. Vi börjar med `0` och lägger till varje värde.

````python
total = 0

for price in prices:
    total = total + price

print(total)
````

## 11. Mönster 3: filtrera

Ibland vill vi skapa en **ny lista** med bara de värden som uppfyller ett villkor. Då börjar vi med en tom lista och använder `append()`.

````python
expensive_prices = []

for price in prices:
    if price > 20:
        expensive_prices.append(price)

print(expensive_prices)
````

## 12. `range()`

`range()` skapar en följd av heltal som kan användas i en loop.

````python
for i in range(5):
    print(i)
````

`range(5)` börjar på 0 och slutar **före** 5.

````python
print(list(range(5)))
print(list(range(1, 6)))
````

När vi redan har en lista är det oftast tydligare att loopa direkt över värdena:

```python
for price in prices:
    print(price)
```

snarare än att skapa index med `range(len(prices))`.

## 13. Sammanhängande exempel

Vi har temperaturer från en vecka. Vi vill:

1. skriva ut varje temperatur,
2. markera temperaturer över 20 grader,
3. räkna antalet varma dagar.

````python
temperatures = [12, 14, 18, 21, 23, 19, 15]
hot_days = 0

for temperature in temperatures:
    print(temperature)

    if temperature > 20:
        print("  varmt")
        hot_days = hot_days + 1

print(f"Antal varma dagar: {hot_days}")
````
