# Iteratorer

En iterator är ett objekt som innehåller ett räknebart antal värden och som kan itereras.

Tekniskt sett, i Python, är en iterator ett objekt som implementerar iteratorprotokollet, vilket består av metoderna __iter__() och __next__()

````python
min_tupel = ("apple", "banana", "cherry")
min_iterator = iter(min_tupel)

for item in min_iterator:
    print(item)
````

````python
#next(min_iterator)
min_iterator = iter(min_tupel) #Måste initialisera igen. Den är annars tom.

print(next(min_iterator))
print(next(min_iterator))
print(next(min_iterator))
````

## Funktioner i klasser

Det finns olika typer av funktioner som kan defineras i klasser, idag tittar vi på den vanligaste typen. Andra ord för en sådan funktion är **medlemsfunktion** (member function), **metod** (method).

Den ser ut som en helt vanlig Pythonfunktion men tar alltid minst ett argument, `self`.

```python
class Greeter:
    def greet(self):
        ...
```

```python
class Greeter:
    def greet(self):
        ...
```

Argumentet `self` innehåller alltid objektet/instansen som funktionen kallades på.
När vi skapar en instans med
```python
g = Greeter()
```
och kallar på metoden greet
```python
g.greet()
```
så kommer Python att översätta detta till
```python
Greeter.greet(g)
```

Argumentet `self` är användbart för det låter oss komma åt variabler och funktioner som hör till samma objekt.

Låt oss skapa en ny klass för att definera en person.

````python
class Person:
    # En metod med två argument: 
    # först instansen (self) och sedan ett namn
    def set_name(self, name):
        # Spara namnet i en medlemsvariabel
        self.name = name
        
    def greet(self):
        print(f"Hi! My name is {self.name}.")
        
````

````python
# Skapa en instans av Person
p1 = Person()
# Kalla på set_name, notera att det bara är ett
# argument här.
p1.set_name("Fredrik")

#Hälsa med funktionen greet
p1.greet()
````

````python
#Man kan somsagt skapa "oändligt" många instanser av typen person.
p1 = Person()
p2 = Person()
p1.set_name("Fredrik")
p2.set_name("Caroline")

p1.greet()
p2.greet()
````

# Åtkomst av medlemsvariabler och funktioner

I Python finns alla medlemsvariabler och funktioner tillgängliga som attribut på en instans.

````python
p = Person()
p.set_name("Hilma")

# Vi kan komma åt medlemsvariabler på instansen direkt
print(p.name)
````

````python
# Byt namn till Alf
p.name = "Alf"
p.greet()
````

Även om det går att komma åt och ändra datan i ett objekt på detta sätt så är det mycket tydligare och enklare att använda om definition och hantering av data sker genom medlemsfunktioner. På så sätt vet vi genom att titta på implementeringen av klassen vad en instans kommer innehålla.

För att underlätta detta yttligare så finns det en speciell funktion, s.k. konstruktor, som körs när ett objekt skapas. Här kan vi skapa alla medlemsvariabler så att de finns tillgängliga från början. I Python heter konstruktorn alltid `__init__`.

```python
class Person:
    def __init__(self, name):
        self.name = name
    ...
```

````python
# Minimalt exempel
class Person:
    def __init__(self, name='123'):
        self.name = name
        
p = Person(name="Louis")
print(p.name)
````

````python
#Notera att man måste sätta name när man initaliserar klassen
p = Person()
````

````python
# Uppdatering till Person klassen med hjälp av konstruktor
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def say_age(self):
        print(f"I am {self.age} years old.")
        
    def say_name(self):
        print(f"My name is {self.name}.")
        
    def greet(self):
        self.say_name()
        self.say_age()
        
````

````python
p1 = Person("Otis", 5)
p2 = Person("Caroline", 59)

p1.greet()
p2.greet()
````

````python
class Aritmetik:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def addition(self):
        return self.a + self.b

    def subtraktion(self):
        return self.a - self.b

    def multiplikation(self):
        return self.a * self.b

    def division(self):
        if self.b == 0:
            return "Kan inte dividera med noll!"
        return self.a / self.b
    
````

````python
# Exempel på hur man använder klassen:
aritmetik_objekt = Aritmetik(10, 5)

print("Addition:", aritmetik_objekt.addition())          # Output: 15
print("Subtraktion:", aritmetik_objekt.subtraktion())    # Output: 5
print("Multiplikation:", aritmetik_objekt.multiplikation())  # Output: 50
print("Division:", aritmetik_objekt.division())          # Output: 2.0
````
