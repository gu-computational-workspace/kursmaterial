# Projekt

I kursen ska du genomföra ett projekt där du använder Python för att undersöka **två frågeställningar**:

1. en frågeställning med **matematisk anknytning**, och
2. en frågeställning med **fysikalisk anknytning**.

Projektet ska visa att du kan använda flera av kursens verktyg för att formulera ett problem, genomföra en beräkning, simulering eller dataanalys, visualisera resultatet och diskutera vad resultatet betyder.


## Projektets mål

Efter projektet ska du kunna visa att du kan:

- formulera en tydlig frågeställning som går att undersöka med Python,
- strukturera kod med variabler, villkor, loopar och **funktioner** när det är lämpligt,
- arbeta med data i listor och/eller **pandas DataFrames**,
- göra relevanta beräkningar och sammanfattningar,
- skapa minst en tydlig **visualisering** för varje frågeställning,
- tolka resultaten och diskutera begränsningar i metoden,
- och, om det passar frågeställningen, använda enkel **maskininlärning** för regression eller klassificering.


## Två mindre undersökningar

Tänk på projektet som två mindre *computational investigations* snarare än två fristående program.

Varje del ska innehålla:

1. **Frågeställning** – Vad vill du undersöka?
2. **Metod** – Hur använder du Python för att undersöka frågan?
3. **Kod** – En tydligt strukturerad lösning.
4. **Resultat** – Beräkningar, sammanfattningar och minst en visualisering.
5. **Tolkning** – Vad visar resultatet?
6. **Begränsningar** – Finns antaganden, förenklingar eller osäkerheter som är viktiga att nämna?

Det är helt okej om den ena delen bygger på simulering och den andra på ett verkligt dataset.


## Minimikrav

Ett färdigt projekt ska innehålla:

- [ ] Två tydliga problemformuleringar: en med matematisk och en med fysikalisk anknytning.
- [ ] Pythonkod som faktiskt används för att undersöka båda frågeställningarna.
- [ ] Minst **en egen funktion** som fyller en tydlig roll i projektet.
- [ ] Minst **en dataanalys eller simulering** där flera observationer/resultat bearbetas.
- [ ] Minst **en visualisering per frågeställning**.
- [ ] Tydliga rubriker, förklarande text och läsbar kod.
- [ ] En kort tolkning av resultaten för båda frågeställningarna.
- [ ] En kort diskussion om metodens begränsningar eller antaganden.

Projektet lämnas i första hand in som en **Jupyter Notebook (`.ipynb`)**, så att text, kod, figurer och resultat kan presenteras tillsammans. Om projektet behöver extra datafiler ska även dessa lämnas in.

## Maskininlärning är valbart

Du **får** använda maskininlärning om frågeställningen lämpar sig för prediktion eller klassificering, men det är inget krav för godkänt projekt.

Om du använder maskininlärning ska du:

- tydligt beskriva vad som är **features (`X`)** och **target (`y`)**,
- dela data i **tränings- och testdata**,
- utvärdera modellen med ett relevant mått, till exempel MAE för regression eller accuracy för klassificering,
- jämföra resultatet med en enkel **baseline** när det är rimligt,
- och kort diskutera modellens begränsningar.

Använd inte maskininlärning bara för att det går. Om en vanlig beräkning, simulering eller statistisk analys svarar bättre på frågan är det oftast det bättre valet.
