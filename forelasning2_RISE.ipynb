# Jämförelser

En jämförelse ger alltid värdet `True` eller `False`.

## Vad tror ni att resultaten blir?


````python
18 < 20
18 > 20
20 == 20
20 != 20
````

# Jämförelseoperatorer

| Operator | Betyder |
|---|---|
| `<` | mindre än |
| `>` | större än |
| `<=` | mindre än eller lika med |
| `>=` | större än eller lika med |
| `==` | lika med |
| `!=` | inte lika med |

**Obs:** `=` tilldelar ett värde. `==` jämför två värden.


# Vad händer vid gränsen?

Vilka av dessa är `True`?


````python
temperature = 20

temperature < 20
temperature <= 20
temperature == 20
````

# `if`: gör något om ett villkor är sant


````python
temperature = 23

if temperature > 20:
    print("Det är varmt")
````

# Indrag är en del av Python

Koden som hör till `if`-satsen måste vara indragen.


````python
temperature = 23

if temperature > 20:
    print("Det är varmt")
    print("Ta med vatten")

print("Programmet fortsätter")
````

# `if` och `else`

Programmet kan välja mellan två alternativ.


````python
temperature = 18

if temperature > 20:
    print("Det är varmt")
else:
    print("Det är inte varmt")
````

# Vad skrivs ut?

Fundera först på tre fall:

- `temperature = 5`
- `temperature = 15`
- `temperature = 25`


````python
temperature = 15

if temperature < 10:
    print("Kallt")
elif temperature < 20:
    print("Lagom")
else:
    print("Varmt")
````

# Ordningen spelar roll

Python provar villkoren **uppifrån och ned** och tar det första som är sant.


````python
temperature = 5

if temperature < 20:
    print("Under 20")
elif temperature < 10:
    print("Under 10")
````

# Flera villkor samtidigt

Vi kan kombinera villkor med:

- `and` – båda måste vara sanna
- `or` – minst ett måste vara sant
- `not` – vänder på `True` / `False`


````python
age = 22
has_ticket = True

if age >= 18 and has_ticket:
    print("Välkommen!")
````

# Vad blir resultatet?


````python
temperature = 18
is_raining = True

print(temperature > 20 and is_raining)
print(temperature > 20 or is_raining)
print(not is_raining)
````

# En liten fråga

Hur skulle ni skriva ett villkor som är sant om temperaturen ligger **mellan 10 och 20 grader**?


````python
temperature = 16

if temperature >= 10 and temperature <= 20:
    print("Mellan 10 och 20 grader")
````

# Paus

När vi kommer tillbaka: **loopar**.


# Varför loopar?

Vi har en lista:


````python
temperatures = [12, 15, 18, 17, 11]
````

Vi skulle kunna skriva:


````python
print(temperatures[0])
print(temperatures[1])
print(temperatures[2])
print(temperatures[3])
print(temperatures[4])
````

# Men om listan innehåller 1 000 värden?

Vi vill uttrycka:

> **Gör samma sak för varje värde i listan.**


# `for`: ett värde i taget


````python
temperatures = [12, 15, 18, 17, 11]

for temperature in temperatures:
    print(temperature)
````

# Läs koden nästan som svenska

```text
för varje temperature i temperatures:
    skriv ut temperature
```

Variabeln `temperature` får ett nytt värde för varje varv.


# Vad skrivs ut?


````python
names = ["Ada", "Grace", "Guido"]

for name in names:
    print("Hej", name)
````

# Loopen kan göra en beräkning


````python
temperatures = [0, 10, 20, 30]

for celsius in temperatures:
    fahrenheit = celsius * 9 / 5 + 32
    print(celsius, "C =", fahrenheit, "F")
````

# Den viktiga kombinationen

## `for` + `if`

Nu kan vi undersöka varje värde och välja vad som ska hända.


````python
prices = [18.95, 19.50, 20.10, 19.85, 21.30]

for price in prices:
    if price > 20:
        print(price)
````

# Vad händer om vi lägger till `else`?


````python
for price in prices:
    if price > 20:
        print(price, "dyrt")
    else:
        print(price, "inte dyrt")
````

# Hur räknar vi antalet?

Vi behöver en variabel som håller reda på hur många vi hittat.

Börja på noll:


````python
count = 0

for price in prices:
    if price > 20:
        count = count + 1

print(count)
````

# Följ värdet på `count`

| Pris | Över 20? | `count` efter varvet |
|---:|:---:|---:|
| 18.95 | nej | 0 |
| 19.50 | nej | 0 |
| 20.10 | ja | 1 |
| 19.85 | nej | 1 |
| 21.30 | ja | 2 |

Detta mönster kallas ofta en **räknare** (counter).


# Vi kan också bygga upp en summa

Vi vet redan att Python har `sum(prices)`.

Men hur skulle vi göra själva?


````python
total = 0

for price in prices:
    total = total + price

print(total)
print(sum(prices))
````

# Ett tredje vanligt mönster: bygg en ny lista

Vi vill spara alla priser över 20 kr.


````python
expensive_prices = []

for price in prices:
    if price > 20:
        expensive_prices.append(price)

print(expensive_prices)
````

# `range()`

Ibland vill vi upprepa något ett bestämt antal gånger.


````python
for i in range(5):
    print(i)
````

# Var börjar `range(5)`?

## Vad tror ni skrivs ut?

`range(5)` ger talen:

**0, 1, 2, 3, 4**

Inte 1 till 5.


````python
list(range(5))
list(range(1, 6))
````

# När vi redan har en lista...

Föredra oftast detta:


````python
for price in prices:
    print(price)
````

...framför detta:


````python
for i in range(len(prices)):
    print(prices[i])
````

Den första versionen säger tydligare vad vi menar:

> för varje pris i listan


# Avslutande problem

Vi har temperaturerna för en vecka:


````python
temperatures = [12, 14, 18, 21, 23, 19, 15]
````

Skriv ett program som:

1. skriver ut varje temperatur
2. skriver `varmt` efter temperaturer över 20 grader
3. räknar hur många dagar som var över 20 grader
4. skriver ut antalet varma dagar sist

## Hur skulle ni börja?


````python
hot_days = 0

for temperature in temperatures:
    print(temperature)

    if temperature > 20:
        print("varmt")
        hot_days = hot_days + 1

print(f"Antal varma dagar: {hot_days}")
````

# En liten utvidgning

Kan vi samtidigt skapa en lista med de varma temperaturerna?


````python
hot_days = 0
hot_temperatures = []

for temperature in temperatures:
    if temperature > 20:
        hot_days = hot_days + 1
        hot_temperatures.append(temperature)

print(f"Antal varma dagar: {hot_days}")
print(f"Varma temperaturer: {hot_temperatures}")
````
