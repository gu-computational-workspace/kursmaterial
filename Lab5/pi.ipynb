# Laboration 5 - Approximation av $\pi$

Det finns [många sätt](https://en.wikipedia.org/wiki/Approximations_of_π) att approximera $\pi$ numeriskt, t.ex med serieutvecklingar.

En enkel metod är att slumpa punkter i planet inom en kvadrat och sedan jämföra hur många som hamnade innanför enhetscirkeln. 

Andelen punkter som hamnar innanför cirkeln borde närma sig $\frac{A_{\text{enhetscirkel}}}{A_{\text{kvadrat}}}$.

Vi har alltså att
$$ \frac{A_{\text{enhetscirkel}}}{A_{\text{kvadrat}}} \approx \frac{N_{inne}}{N_{total}} $$
där $N_{inne}$ är antalet punkter som hamnade innanför enhetscirkeln av totalt $N_{total}$ slumpade punkter.

Eftersom area av enhetscirkeln och kvadraten är
$$ A_{\text{enhetscirkel}} = \pi $$
$$ A_{\text{kvadrat}} = 4 $$

så är en approximation av $\pi$ därför
$$
 \pi \approx 4 \cdot \frac{N_{inne}}{N_{total}}
 $$
