## Kan vi återanvända kod i stället för att skriva om den?

# Ett exempel
Vi vill räkna ut medelvärdet för en lista.

````python
temperatures = [12, 17, 21, 23, 18]

average = sum(temperatures) / len(temperatures)
print(average)
````

# Och för en annan lista?

Vi kan förstås skriva samma typ av kod igen.

````python
week_1 = [12, 17, 21, 23, 18]
week_2 = [8, 10, 11, 15, 13]

average_1 = sum(week_1) / len(week_1)
average_2 = sum(week_2) / len(week_2)

print(average_1)
print(average_2)
````

# Funktioner

En **funktion** är ett namngivet kodblock som vi kan använda flera gånger.

Vi skapar en funktion med `def`.

````python
def average(values):
    return sum(values) / len(values)
````

# Anropa funktionen

Nu kan samma kod användas för olika listor.

````python
week_1 = [12, 17, 21, 23, 18]
week_2 = [8, 10, 11, 15, 13]

print(average(week_1))
print(average(week_2))
````

# Läs funktionen som en liten maskin

```text
värde in  →  funktion  →  värde ut
```

I funktionen

```python
def average(values):
    return sum(values) / len(values)
```

är `values` namnet på värdet som kommer **in**.

# Parameter och argument

Orden dyker ofta upp i dokumentation:

```python
def average(values):
    ...
```

`values` är en **parameter**.

När vi skriver `average(week_1)` är `week_1` ett **argument**.

Det viktigaste är idén: **funktionen får ett värde att arbeta med.**

# Ett enkelt exempel

## Vad tror ni att anropen ger?

````python
def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

print(celsius_to_fahrenheit(0))
print(celsius_to_fahrenheit(20))
print(celsius_to_fahrenheit(100))
````

# En funktion kan ha flera parametrar

````python
def rectangle_area(width, height):
    return width * height

print(rectangle_area(3, 4))
print(rectangle_area(10, 2))
````

# `return` är inte samma sak som `print()`

Två funktioner kan se nästan likadana ut, men bete sig olika.

## Vad tror ni att `result_1` och `result_2` blir?

````python
def double_print(x):
    print(x * 2)

def double_return(x):
    return x * 2

result_1 = double_print(10)
result_2 = double_return(10)

print("result_1:", result_1)
print("result_2:", result_2)
````

# När använder vi `print()` och när använder vi `return`?

- `print()` **visar** något för användaren.
- `return` **skickar tillbaka** ett värde från funktionen.

Ett returvärde kan sparas i en variabel och användas vidare.

````python
def double(x):
    return x * 2

answer = double(10)
final_answer = answer + 5

print(final_answer)
````

# Funktioner kan innehålla `if`

Vi kan återanvända allt vi redan kan inuti en funktion.

````python
def describe_temperature(temp):
    if temp < 10:
        return "kallt"
    elif temp < 20:
        return "lagom"
    else:
        return "varmt"
````

# Vad returneras?

Fundera först på:

- `describe_temperature(5)`
- `describe_temperature(15)`
- `describe_temperature(25)`

````python
print(describe_temperature(5))
print(describe_temperature(15))
print(describe_temperature(25))
````

# Funktioner och loopar tillsammans

````python
temperatures = [7, 12, 18, 24]

for temp in temperatures:
    description = describe_temperature(temp)
    print(temp, description)
````

# Från labb 2: nästan samma kod två gånger

Vi ville till exempel räkna namn som började på olika bokstäver.

Det är ett bra tillfälle att skapa en funktion.

# Gör det till en funktion

Vad behöver variera mellan gångerna?

````python
def count_names_starting_with(names, letter):
    count = 0

    for name in names:
        if name[0] == letter:
            count += 1

    return count
````

# Samma funktion – olika frågor

````python
names = ["Anna", "Tobias", "Maja", "Tove", "Mohammed", "Alva"]

print(count_names_starting_with(names, "T"))
print(count_names_starting_with(names, "M"))
print(count_names_starting_with(names, "A"))
````

# En liten förbättring med strängmetoder

Strängar har färdiga metoder som kan vara praktiska.

`startswith()` frågar om en sträng börjar på en viss text.

````python
name = "Tobias"

print(name.startswith("T"))
print(name.startswith("M"))
````

# Samma funktion med `startswith()`

````python
def count_names_starting_with(names, letter):
    count = 0

    for name in names:
        if name.startswith(letter):
            count += 1

    return count
````

# Dictionary

En **dictionary** lagrar värden med **nycklar**.

I stället för att fråga efter position `1` kan vi fråga efter `"age"`.

````python
person = {
    "name": "Ada",
    "age": 28,
    "city": "Göteborg"
}

print(person["name"])
print(person["age"])
print(person["city"])
````

# Lista eller dictionary?

**Lista**

```python
temperatures = [12, 17, 21]
```
Positionen är viktig.

**Dictionary**

```python
person = {"name": "Ada", "age": 28}
```
Nyckeln beskriver vad värdet betyder.

# Tillbaka till namnlabben

Vi kunde enkelt räkna namn som började på **en viss bokstav**.

Men tänk om vi vill räkna **alla begynnelsebokstäver samtidigt**?

Vi behöver lagra något i stil med:

```text
A → 2
M → 2
T → 2
...
```

Det är precis vad en dictionary passar för.

# Börja med en tom dictionary

````python
counts = {}

names = ["Anna", "Tobias", "Maja", "Tove", "Mohammed", "Alva"]
````

# Ett namn i taget

För varje namn tar vi första bokstaven.

````python
for name in names:
    letter = name[0]
    print(name, letter)
````

# Men första gången finns bokstaven inte ännu

Vi kan kontrollera det med `not in`.

````python
counts = {}

for name in names:
    letter = name[0]

    if letter not in counts:
        counts[letter] = 0

    counts[letter] += 1

print(counts)
````

# Följ vad som händer

Med namnen

```python
["Anna", "Tobias", "Maja", "Tove", "Mohammed", "Alva"]
```

byggs dictionaryn steg för steg:

```text
Anna      → {'A': 1}
Tobias    → {'A': 1, 'T': 1}
Maja      → {'A': 1, 'T': 1, 'M': 1}
Tove      → {'A': 1, 'T': 2, 'M': 1}
Mohammed  → {'A': 1, 'T': 2, 'M': 2}
Alva      → {'A': 2, 'T': 2, 'M': 2}
```

# Nu går det att ställa nya frågor

````python
print("A:", counts["A"])
print("M:", counts["M"])
print("T:", counts["T"])
````

# Och nu: kombinera allt

Kan vi göra en **funktion** som tar en lista med namn och returnerar en dictionary med antal per begynnelsebokstav?

# Ett komplett exempel

````python
def count_initials(names):
    counts = {}

    for name in names:
        letter = name[0]

        if letter not in counts:
            counts[letter] = 0

        counts[letter] += 1

    return counts
````

# Funktionen kan återanvändas

````python
names_1 = ["Anna", "Tobias", "Maja", "Tove", "Mohammed", "Alva"]
names_2 = ["Luna", "Leo", "Alice", "Liv", "Noah"]

print(count_initials(names_1))
print(count_initials(names_2))
````
