## 1. Varför funktioner?

Efter föreläsning 2 kan vi skriva ganska långa kodblock. Problemet är att samma idé lätt behöver användas flera gånger.

````python
week_1 = [12, 17, 21, 23, 18]
week_2 = [8, 10, 11, 15, 13]

average_1 = sum(week_1) / len(week_1)
average_2 = sum(week_2) / len(week_2)

print(average_1, average_2)
````

## 2. Första funktionen

En funktion definieras med `def`. Koden körs inte när funktionen definieras; den körs först när funktionen **anropas**.

````python
def average(values):
    return sum(values) / len(values)
````

````python
week_1 = [12, 17, 21, 23, 18]
week_2 = [8, 10, 11, 15, 13]

print(average(week_1))
print(average(week_2))
````

Tänk på funktionen som en liten maskin:

**input → arbete → output**

Här är `values` input och `return` skickar tillbaka output.

## 3. Parameter och argument

I definitionen:

```python
def celsius_to_fahrenheit(celsius):
```

är `celsius` en **parameter**.

I anropet:

```python
celsius_to_fahrenheit(20)
```

är `20` ett **argument**.

````python
def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

print(celsius_to_fahrenheit(0))
print(celsius_to_fahrenheit(20))
print(celsius_to_fahrenheit(100))
````

## 4. `return` är inte samma sak som `print()`

`print()` visar något för användaren. `return` skickar ett värde tillbaka från funktionen så att resten av programmet kan använda det.

````python
def double_print(x):
    print(x * 2)

def double_return(x):
    return x * 2

a = double_print(10)
b = double_return(10)

print("a:", a)
print("b:", b)
````

`double_print()` skriver ut 20 men returnerar inget explicit, så `a` blir `None`.

`double_return()` returnerar 20, så värdet kan sparas i `b` och användas senare.

## 6. Funktioner och loopar tillsammans

En funktion kan också loopa över en lista.

````python
def count_names_starting_with(names, letter):
    count = 0

    for name in names:
        if name.startswith(letter):
            count += 1

    return count

names = ["Anna", "Tobias", "Maja", "Tove", "Mohammed", "Alva"]
print(count_names_starting_with(names, "T"))
print(count_names_starting_with(names, "M"))
````

Det här är ett viktigt motiv för funktioner: i stället för att skriva nästan samma kod en gång för T och en gång för M har vi skrivit **en generell lösning**.

## 7. Lokala variabler

Variabler som skapas inne i en funktion är normalt **lokala** för funktionen.

````python
def square(x):
    result = x * x
    return result

print(square(4))
````

Variabeln `result` används inne i funktionen. Vi behöver normalt inte, och bör ofta inte, bygga funktioner som är beroende av en massa globala variabler.

## 8. Från listor till dictionaries

En lista är bra när **positionen** betyder något.

````python
person = ["Ada", 28, "Göteborg"]

print(person[0])
print(person[1])
````

Men efter ett tag blir `person[1]` svårt att läsa. Vad betyder index 1?

En **dictionary** kopplar i stället en nyckel till ett värde.

````python
person = {
    "name": "Ada",
    "age": 28,
    "city": "Göteborg"
}

print(person["name"])
print(person["age"])
````

## 9. Ändra och lägga till värden

## 10. Loopa över en dictionary

En vanlig loop över en dictionary ger nycklarna.

````python
for key in person:
    print(key, person[key])
````

Med `.items()` får vi nyckel och värde samtidigt.

````python
for key, value in person.items():
    print(key, value)
````

## 11. Dictionary som räknare

Nu återkommer vi till namnproblemet från laboration 2. Vi vill räkna hur många namn som börjar på varje bokstav.

Vi vet inte i förväg vilka bokstäver som dyker upp, så en dictionary passar bra.

````python
names = ["Anna", "Tobias", "Maja", "Tove", "Mohammed", "Alva"]
counts = {}

for name in names:
    letter = name[0]

    if letter not in counts:
        counts[letter] = 0

    counts[letter] += 1

print(counts)
````

Läs logiken stegvis:

1. Ta första bokstaven.
2. Om bokstaven inte finns i dictionaryn: skapa den med värdet 0.
3. Öka räknaren med 1.

## 12. Kombinera funktion och dictionary

Nu kapslar vi in samma idé i en återanvändbar funktion.

````python
def count_initials(names):
    counts = {}

    for name in names:
        letter = name[0]

        if letter not in counts:
            counts[letter] = 0

        counts[letter] += 1

    return counts

print(count_initials(names))
````
