## Här kommer en ordlista med termer och begrepp relaterade till **neurala nätverk** och deras användning inom **maskininlärning**:

Denna ordlista täcker de grundläggande begreppen och teknikerna inom neurala nätverk och maskininlärning.

---

### **A**
- **Aktiveringsfunktion (Activation Function)**: En funktion som appliceras på varje neurons utdata för att införa icke-linjäritet i nätverket. Vanliga aktiveringsfunktioner är **ReLU**, **Sigmoid**, **Tanh** och **Softmax**.
- **Autoencoder**: En typ av neuralt nätverk som lär sig en komprimerad representation av data, ofta använd för dimensionreduktion eller anomalidetektion.

---

### **B**
- **Backpropagation**: En algoritm för att uppdatera vikterna i ett neuralt nätverk genom att beräkna och sprida fel (gradienter) bakåt genom nätverket för att minimera förlusten.
- **Batch**: En uppsättning datapunkter som bearbetas samtidigt under träning av nätverket.
- **Batch Normalization**: En teknik för att normalisera aktiveringar mellan lagren, vilket förbättrar nätverkets stabilitet och träningshastighet.

---

### **C**
- **CNN (Convolutional Neural Network)**: En typ av neuralt nätverk som används för att behandla bilddata genom att använda konvolutionella lager för att extrahera funktioner från bilder.
- **Cross-Entropy Loss**: En förlustfunktion som mäter skillnaden mellan den verkliga fördelningen och den predicerade sannolikheten för varje klass i ett klassificeringsproblem.

---

### **D**
- **Dropout**: En regulariseringsteknik där slumpmässigt valda neuroner stängs av under träning för att förhindra överanpassning.
- **Deep Learning**: En typ av maskininlärning där djupa neurala nätverk med många lager används för att modellera komplexa relationer i data.
- **Dense Layer**: Ett lager i ett neuralt nätverk där varje neuron är ansluten till alla neuroner i föregående lager.

---

### **E**
- **Epoch**: Ett genomlopp av hela träningsdatan. Under varje epoch uppdateras nätverkets vikter för att minska förlusten.
- **Embedding**: En kontinuerlig representation av kategoriska eller sekventiella data (t.ex. ord i en text) i ett högdimensionellt rum.

---

### **F**
- **Feedforward Neural Network**: Ett typ av neuralt nätverk där informationen flödar i en riktning från indata till utdata utan återkoppling.
- **Funktionell API**: En metod för att definiera neurala nätverk där lager kan kopplas ihop på ett mer flexibelt sätt än i det traditionella "Sequential API".

---

### **G**
- **Gradient Descent**: En optimeringsmetod som används för att minimera förlusten genom att justera nätverkets vikter i riktning mot den negativa gradienten av förlustfunktionen.
- **Generative Adversarial Network (GAN)**: Ett typ av neuralt nätverk som består av två delar: en **generator** som skapar data och en **diskriminator** som försöker skilja mellan verkliga och genererade data.

---

### **H**
- **Hidden Layer**: Ett lager mellan indata och utdata i ett neuralt nätverk. Dessa lager bearbetar och extraherar funktioner från data.
- **Hyperparameter**: Parametrar som styr träningsprocessen, såsom inlärningshastighet, batchstorlek eller antalet lager i nätverket.

---

### **I**
- **Input Layer**: Det första lagret i ett neuralt nätverk som tar emot indata.
- **Inference**: Processen där ett tränat neuralt nätverk används för att göra förutsägelser på nya, osedda data.

---

### **L**
- **Loss Function**: En funktion som mäter skillnaden mellan nätverkets prediktioner och det sanna resultatet. Exempel inkluderar **Mean Squared Error (MSE)** och **Cross-Entropy Loss**.
- **LSTM (Long Short-Term Memory)**: En typ av Recurrent Neural Network (RNN) som är designad för att hantera långsiktiga beroenden och sekventiella data, som tidsserier och text.

---

### **M**
- **Model**: Den matematiska representationen av en funktion som förutsäger utdata från indata, bestående av vikter och aktiveringsfunktioner.
- **Momentum**: En optimeringsteknik som gör det möjligt för gradient descent att röra sig snabbare genom plana områden och minska oscillationer genom att använda tidigare gradienter.
- **Minibatch**: En liten delmängd av hela träningsdatan som används i varje iteration av gradient descent.

---

### **N**
- **Neurons (Neuroner)**: De grundläggande byggstenarna i ett neuralt nätverk. Varje neuron tar emot input, bearbetar det och skickar vidare en output till nästa lager.
- **NLP (Natural Language Processing)**: Ett område inom AI som fokuserar på att bearbeta och analysera text och tal, ofta med hjälp av neurala nätverk som **RNN** eller **Transformer**.

---

### **O**
- **Overfitting**: När en modell presterar bra på träningsdatan men dåligt på nya, osedda data eftersom den har lärt sig specifika detaljer och brus från träningsdatan istället för generaliserbara mönster.
- **Optimizer**: Algoritmer som används för att uppdatera modellens vikter, exempelvis **SGD**, **Adam**, eller **RMSprop**.

---

### **P**
- **Pooling Layer**: Ett lager i ett CNN som minskar dimensionerna av data genom att till exempel använda max-pooling eller genomsnittspooling för att extrahera viktiga funktioner.
- **Perceptron**: Den enklaste typen av neuralt nätverk, med endast ett lager av neuroner.
- **Preprocessing**: Förbehandling av data innan den används för träning, som att normalisera värden eller konvertera text till numeriska representationer.

---

### **R**
- **ReLU (Rectified Linear Unit)**: En aktiveringsfunktion som är mycket populär, definierad som \(f(x) = \max(0, x)\). Den hjälper till att bekämpa problemet med **vanishing gradient**.
- **RNN (Recurrent Neural Network)**: Ett typ av neuralt nätverk som används för sekventiell data, som text eller tidsserier, där nätverket har en form av "minne" av tidigare data.
- **Regularization**: Tekniker för att förhindra överanpassning, som **L2-regularisering**, **dropout**, eller **early stopping**.

---

### **S**
- **Softmax**: En aktiveringsfunktion som används i slutet av nätverk för flervalsklassificering, där den omvandlar utdata till sannolikheter för varje klass.
- **Stochastic Gradient Descent (SGD)**: En variant av gradient descent som uppdaterar vikterna med en slumpmässig underuppsättning (minibatch) av träningsdatan i varje iteration.

---

### **T**
- **Training Data**: Datan som används för att träna modellen. Den används för att justera nätverkets vikter så att modellen kan lära sig.
- **Transfer Learning**: Tekniken att använda en redan tränad modell för ett liknande problem och finjustera den för ett specifikt mål.
- **Tuning**: Justering av modellens hyperparametrar för att förbättra prestanda, t.ex. genom **grid search** eller **random search**.

---

### **U**
- **Underfitting**: När en modell inte lär sig tillräckligt från träningsdatan och presterar dåligt både på träningsdatan och nya data.
- **Unsupervised Learning**: Maskininlärning där modellen inte får märkta data och istället försöker hitta mönster eller grupperingar i datan, t.ex. med **clustering** eller **autoencoders**.

---

### **V**
- **Validation Data**: Datan som används för att utvärdera modellens prestanda under träning och hjälpa till att justera hyperparametrarna.
- **Vanishing Gradient Problem**: Ett problem som uppstår när gradienten försvinner under backpropagation, vilket gör att träning av djupa nätverk blir mycket långsamt eller omöjligt.

---

### **W**
- **Weight**: Parametrar i nätverket som styr styrkan på kopplingarna mellan neuroner i olika lager.
- **Word Embeddings**: Representationer av ord i ett kontinuerligt vektorrum, ofta lärda med metoder som **Word2Vec** eller **GloVe**, för att fånga ordens semantiska betydelse.

---

### **Z**
- **Zero Padding**: Tekniken att lägga till nollor runt kanterna på en bild eller sekvens innan bearbetning, ofta använd i CNN för att bevara dimensionerna efter konvolution.
