# Projekt - maskininlärning med Python

Under kursens gång har vi sett att maskinlärning med neurala nät kan appliceras till nästan vad som helst, allt från funktionsanpassning i en dimension till bildigenkänning.

Nu är det din tur att välja ett problem där maskininlärning kan användas:

### Välj ett problem
* Välj ett problem som du kan applicera maskininlärning på. Det kan vara ett klassificeringsproblem, regressionsproblem eller något annat som du tycker är intressant.
* Hitta data som du kan använda för att träna din maskininlärningsmetod med eller skapa din egen syntetiska data (se föreläsningarna för exempel).
* Implementera ett program som tränar och evaluerar din modell.
* Fundera på om du kan hitta en problemställning som du verkligen vill ha svar på, kanske kan det vara något på ditt jobb som kan optimeras, eller kanske har du en hobby där data finns tillgänglig.

Du får gärna tänka fritt kring möjliga variationer men formatet skall motsvara nedanstående. (Hör gärna av dig om du har en idé som går utanför ramarna här så kan vi prata om hur det lämpar sig.)

## Format för inlämning
* Ett fristående program, Jupyter notebook eller en kombination av dessa som går att köra i miljön på JupyterHub för kursen.
* Data för träning och evaluering: 
    * Om du skapar syntetisk data så skall funktioner för att generera datan ingå i programmet.
    * Om du använder ett publikt (se nedan) dataset så räcker det med att ha med en länk för nedladdning.
    * Om det INTE är ett publikt dataset så skall datan finnas med i inlämningen eller ligga på din JupyterHub. (*Obs max 1GB lagring per användare*)
* **Sammanställning av problemformulering och resultat**
    * Ett dokument som beskriver: 
        * Problemformuleringen: Vad försöker du modellera?
        * Data: Vilket format är datan i? Hur mycket data finns tillgänglig? Visa exempel på träningsdatan.
        * Modell: Hur ser din modell ut?
        * Resultat: Gick det att hitta en bra modell? Visa exempel på prediktioner.
        * Visualisering: Visualisera problemet/datan/resultat, din sammanställning måste innehålla minst en grafisk representation skapad i Python.
    * Om du väljer att använda Jupyter notebooks så är det helt ok att blanda implementering och sammanställning men du måste se till att din notebook går att köra från början till slut och att du lämnar in tillsammans med en PDF-exportering.
    * Det finns inget direkt krav på antalet sidor, men en lämplig omfattning är någonstans 1-5 sidor.
    * Inlämning av projektet sker på Canvas, dvs. skicka in filerna där. Detta gäller koderna med ändelse(`.ipynb` eller `.py`) samt `.pdf`-dokumentet. Om datan du använder är över 20-30mb men mindre än 1gb, ha den helst på er JupyterHub.
        
## Publika dataset

Det finns många dataset tillgängliga på internet, en bra källa är [Kaggle](https://www.kaggle.com/datasets). Här kan du söka bland tusentals olika problem med tillhörande data. 
Testa t.ex att söka på "classification" eller "regression" för att hitta möjliga klassificeringsproblem eller regressionsproblem.

Du får också använda andra källor som tillhandahåller rådata, t.ex [SCB](https://scb.se) eller [SMHI](https://smhi.se).

Du får använda de dataset vi har jobbat med i kursen, t.ex MNIST samt FashionMNIST, men din problemformulering måste tillföra något från analysen vi redan gjort. T.ex så kan "utmaningarna" från laboration 3 vara en grund för ett projekt.
